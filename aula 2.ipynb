```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv('pedidos_narashop.csv', sep=';')
df
```

```python
df['valor_pedido'] = df['quantidade'] * df['valor_unitario']
df
```

```python
df_fatura_categoria = df.groupby('categoria')['valor_pedido'].sum().reset_index().sort_values(by='valor_pedido', ascending=False)


df_fatura_categoria
```

```python
df_3 = df.loc[(df['status_pedido'] == 'Entregue') & (df['valor_pedido'] > 500)]
df_3
```

```python
df_produto_valor_vendido = df.groupby('produto')['valor_pedido'].sum().reset_index().sort_values(by='valor_pedido', ascending= False)


df_produto_valor_vendido.head()
```

```python
array_produto_val_pedido = np.array(df_produto_valor_vendido['valor_pedido'])
```

```python
media = np.mean(array_produto_val_pedido)
mediana = np.median(array_produto_val_pedido)
q1 = np.percentile(array_produto_val_pedido, 25)
q3 = np.percentile(array_produto_val_pedido, 75)
distancia = round(((media - mediana) / mediana) * 100,2)
iqr = q3 - q1
limite_superior = q3 + (1.5 * iqr)
limite_inferior = q1 - (1.5 * iqr)

print (f'Q1 {q1}')
print (f'Q3 {q3}')
print (f'Média {media}')
print (f'Mediana {mediana}')
print (f'Distancia {distancia:.2%}')
print(f'Limite Superior {limite_superior}')
print(f'Limite Inferior {limite_inferior}')
```

```python
prod_outlier_sup = df_produto_valor_vendido.loc[df_produto_valor_vendido['valor_pedido'] > limite_superior]
prod_outlier_sup
```

```python
prod_outlier_inf = df_produto_valor_vendido.loc[df_produto_valor_vendido['valor_pedido'] < limite_inferior]
prod_outlier_inf
```

```python
top_25_produto = df_produto_valor_vendido.loc[df_produto_valor_vendido['valor_pedido'] > q3]
top_25_produto
```

```python
top_menos_25_produto = df_produto_valor_vendido.loc[df_produto_valor_vendido['valor_pedido'] < q1]
top_menos_25_produto.sort_values(by='valor_pedido')
```
