```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
categorias = ['Eletrônicos', 'Roupas', 'Alimentos', 'Casa']
valores = [4200, 2800, 3600, 1900]

dados = {'categorias' : ['Eletrônicos', 'Roupas', 'Alimentos', 'Casa'], 'valores' : [4200, 2800, 3600, 1900]}
df = pd.DataFrame(dados)
df
```

```python
#trabalhando com gráfico. a biblioteca matplot permite isso
plt.bar(df['categorias'], df['valores'])
plt.title('Total de Vendas por Categoria - NaraStore')
plt.show()
```

```python
df = pd.read_csv('vendas_narastore.csv')
df
```

```python
df_cat = df.groupby('categoria')['total_venda'].sum().reset_index().sort_values(by='total_venda', ascending=True) #ascending=True, vai fazer o gráfico ficar cresente
df_cat
```

```python
plt.figure(figsize=(8,7))
plt.bar(df_cat['categoria'], df_cat['total_venda'], color='#FF9272')
plt.title('Total de Vendas por Categoria - NaraStore')
plt.xticks(rotation=22)
plt.show()
```

```python
plt.barh(df_cat['categoria'], df_cat['total_venda'], color='black')
plt.title('Total de Vendas por Categoria - NaraStore')
plt.show()
```
