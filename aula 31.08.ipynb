```python
import pandas as pd
import numpy as np
```

```python
dados = {
'Aluno': ['Ana','Bruno', 'Carol', 'Diego', 'Elisa', 'Felipe', 'Gabi', 'Hugo'],
'Nota': [8.5, 6.0, 9.2, 5.5, 7.8, 4.0, 9.0, 6.5],
'horas de estudo': [6, 3, 8, 2, 5, 1, 7, 4]}

df= pd.DataFrame(dados)
df
```

```python
array = np.array(df['Nota'])
array_horas = np.array(df['horas de estudo'])
```

```python
media = np.mean(array)
mediana = np.median(array)
q1 = np.percentile(array, 25)
q2 = np.percentile(array, 50)
q3 = np.percentile(array,75)
iqr = q3 - q1
limite_superior = q3 + (1.5 * iqr)
limite_inferior = q1 - (1.5 * iqr)
distancia = (media - mediana) / mediana

print (f'Q1 {q1}')
print (f'Q2 {q2}')
print (f'Q3 {q3}')
print (f'Média {media:.2f}')
print (f'Mediana {mediana}')
print (f'Distancia {distancia:.2%}')
print(f'Limite Superior {limite_superior}')
print(f'Limite Inferior {limite_inferior}')
```

```python
#variância das notas
variancia = np.var(array)
print (f'variância: {variancia:.2f}')
```

```python
# desvio padrão das notas
desvio = np.std(array)
print(f'Desvio padrão: {desvio:.2f}')
#saída: Desvio padrão: 1.74
#Interpretando com a média
media = np.mean(array)
print(f'Média das notas: {media:.2f}')
print(f'Faixa típica: {media-desvio:.2f} até {media+desvio:.2f}')
```

```python
matriz_cov = np.cov(array, array_horas)
print(f'matriz {matriz_cov}')

cov_valor = matriz_cov[0,1]
print(f'Covariância: {cov_valor:.2f}')
```

```python
print(df[['Nota', 'horas de estudo']].cov())
```

```python
r = np.corrcoef(array, array_horas)
print (f'Pearson r = {r[0,1]:.4f}')
```

```python
corr_pandas = df[['Nota', 'horas de estudo']].corr()
print(corr_pandas)
```

```python
import matplotlib.pyplot as plt
plt.scatter(array_horas, array, color='red', edgecolor='black', s=80)
plt.title('Horas de Estudo vs. Nota')
plt.xlabel('Horas de Estudo')
plt.ylabel('Nota')
plt.show()
```
